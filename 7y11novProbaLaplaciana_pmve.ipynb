# **PROBABILIDAD LAPLACIANA**

La **probabilidad laplaciana** surgió gracias al matemático francés *Pierre Simon Laplace*, a finales del siglo XVIII y principios del XIX. Él desarrolló y formalizó el estudio de la probabilidad, estableciendo que esta podía calcularse dividiendo los casos favorables entre el total de casos posibles, siempre que todos tuvieran la misma posibilidad de ocurrir. Su trabajo fue fundamental para el desarrollo de la teoría de la probabilidad moderna.

Para un experimento aleatorio, se tiene que $\Omega$ es el espacio muestral (el conjunto de posibles resultados).

Sea $A$ un evento, entonces definimos la probabilidad de $A$ como sigue:

$$ \mathbb{P}(A) = \frac{|A|}{|\Omega|}$$

donde $||$ = número de elementos

# **Ejemplo.**
Sea el experimento de lanzar una moneda 3 veces, de manera independiente, es decir, el $i$-ésimo lanzamiento no afecta al $i+1$ lanzamiento. Entonces el espacio muestral asociado a este experimento es:

$$ \Omega = \{A, S\}\times\  \{A, S\} \times\ \{A, S\} = \{A, S\}^3 $$

donde

$A = \{\text{águilas}\}, A = \{\text{soles}\}$ y sabemos que $|\Omega|= 2^3 = 8$.

Sean los eventos:

- $A$ : todos los resultados son águila
- $S$ : todos los resultados son sol
- $E$: por lo menos un resultado es sol (1 o 2 o 3 soles)

Calcular las probabilidades por medio de la probabilidad Laplaciana.

Para hacer esto, vamos a descomponer al evento $A$ de la siguiente manera:

$$ A = A_1 \cap A_2 \cap A_3 $$
donde $A_i$ es que caiga águila en el $i$-ésimo lanzamiento.






Creando el espacio muestral con intertools

```python
# Importamos el módulo
import itertools
from itertools import product
```

```python
# Creando Omega
Omega = set(product({'A','S'}, repeat = 3)) # repeat porque se están lanzando 3 veces una moneda
Omega # set convierte un conjunto y product es el producto cartesiano
```

```python
# La cardinalidad de Omega es
len(Omega)
```

Vamos a crear el evento $A$

```python
# Para A_1: el primer elemento es águila (águila en la primera posición)
A_1 = {om for om in Omega if om[0] == 'A'}
A_1
```

```python
# Para A_2: el segundo elemento es águila (águila en la segunda posición)
A_2 = {om for om in Omega if om[1] == 'A'}
A_2
```

```python
# Para A_3: el tercer elemento es águila (águila en la tercera posición)
A_3 = {om for om in Omega if om[2] == 'A'}
A_3
```

```python
# Intersectar A_1, A_2, y A_3
A = A_1.intersection(A_2.intersection(A_3))
A
```

```python
card_A = len(A)
```

```python
P_A = card_A/len(Omega)
P_A
```

Si queremos el resultado en fracción, utilizar *fractions*

```python
from fractions import Fraction
```

```python
# Crer la función de probabilidad
def p(E, Omega):
  p = Fraction(len(E), len(Omega))
  return p
```

```python
Pro_A = p(A, Omega)
Pro_A
```

```python
# Para forzar a que se escriba de manera más amigable
print(Pro_A)
```

# **Probabilidad Condicional**

Cuando tengo información adicional del experimento.

Si añadimos información o condiciones a nuestro experimento aleatorio, entonces, la probabilidad de un evento cambia.

Por ejemplo, definamos la **probabilidad condicional** de que ocurra un evento $A$ sabiendo que ocurre un evento $B$ de probabilidad positiva, entonces:

$$ \mathbb{P}(A \mid B) = \frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)} = \frac{|A \cap B|}{|B|} $$

```python
# Vamos a definir la función de probabilidad condicional

def prob_cond(E,F):
  return len(E & F) / len(F)
```

Queremos calcular la probabilidad condicional de los siguientes eventos:

- $B$: el primer lanzamiento es águila
- $C$: dos de los resultados son águila (exactamente)

Calcular $\mathbb{P}(B\mid C)  $

```python
# Para el evento B: el primer lanzamiento es águila
B = {om for om in Omega if om[0] == 'A'} # llaves es conjunto {}
    # if es la condicion
    # 0 es el primer resultado, que va a ser guila
B

# Restringe a que el primero sea Águila
```

```python
# Para el evento C: los de los resultados son águila (exactamente)
C = {om for om in Omega if om.count('A') == 2}
C
```

```python
# Calcular P(B|C)
prob_cond(B,C)

# Si de 2 águilas, cuál es la probabilidad de que la primera salga águila
```

# **Independencia de Eventos**

Decimos que dos eventos $A$ y $B$ son independientes si

$$ \mathbb{P}(A\cap B) = \mathbb{P}(A) \mathbb{P}(B) $$



```python
# Definir la función de probabilidad de un evento
def prob(E): # E es evento
    return len(E)/len(Omega)
```

```python
# Definir la función de independencia de eventos
def indep(E,F):
  return prob(E & F) == prob(E) * prob(F)
```

```python
# ¿Son B y C eventos independientes?
indep(B,C)
```
