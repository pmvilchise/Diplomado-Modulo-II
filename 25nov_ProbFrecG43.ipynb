# **Probabilidad Frecuentista**

**Jacob Bernoulli** en 1713 fórmulo y demostró una versión temprana de la ley de los grandes numeros.

Sea un espacio muestral (conjunto finito) podemos definir la probabilidad aosciada a un evento $A$ donde tengo un experiemento aleatorio que se realiza $n$ veces

$$ f_n(A) = \frac{n(A)}{n} $$
donde $n(A)$ es el número de veces en que ocurre $A$.



Sea el experimento aleatorio de lanzar una moneda (no junta).

```python
import random # Para generar números aleatorios
```

```python
def volado(p):
  return 'sol' if random.random() < p else 'aguila'
# random.random() genera un número entre 0 y 1
```

```python
# 40% de obtener sol
volado(0.4)
```

* Vamos a realizar 10 volados con probabilidad p

$\color{pink}{Nota.}$ El resultado va a cambiar cada vez que se ejecute, ya que son números aleatorios

```python
# range() crea un contador de 0 a un cierto número
vol = [volado(0.4) for i in range(10)]
vol
```

```python
# Contar el número de soles en una lista
vol.count('sol')
```

```python
# Contar el número de águilas en una lista
vol.count('aguila')
```

$\color{pink}{Ejemplo.}$

Vamos a simular la probabilidad frecuentista, es decir, vamos a realizar un experimento con 1000 volados (individual) y puedo repetirli 500 veces (500 personas)

```python
# Crear una lista para almacenar probabilidades
p = []
M = 1000 # Número de experimentos (personas)
N = 1000 # cada experimento tiene N volados

for i in range(M):
    resultados = [] # para almacenar los resultados

    for j in range(N):
      resultado = volado(0.5) # la probabilidad se puede cambiar a criterio propio o lo que se solicite
      resultados.append(resultado)

    n_soles = resultado.count('sol')

    # Probabilidad frecuentista
    p_soles = n_soles / N # probabilidad de obtener soles

    # Añadir estas probabilidades a la lista p
    p.append(p_soles)

# Probabilidad frecuentista promedio
p_soles_prom = sum(p)/ 500 # suma de probabilidades / No. personas
p_soles_prom
```

# **Probabilidad Geométrica**

## $\color{pink}{Ejemplo.}$ Problema del encuentro

Dos personas planean encontrarse en un lugar especifico, en un rango de una hora $[0,60]$ minutos (1 hora).
 - Si A llega primero, entonces piensa esperar 10 minutos.
 - Si B llega primero, entonces piensa esperar 10 minutos

Se elige un número al azar entre $[0,60]$ y supongamos que el par de números se elige sobre el cuadrado $[0,60]\times[0,60]$ de modo que la proobabilidad de un evento sea el área favorable entre el área total.

Sea

- $A$: las dos personas se encuentran a una distancia de 10 minutos (es decir, $|x-y|\le 10$.

Nos gustaría encontrar la probabilidad de $A$.



* Simular:
Muestra de 1000 simulaciones

```python
import numpy as np
import matplotlib.pyplot as plt

# Generar 1000 simulaciones
N = 1000000

# Generar los tiempos aleatorios entre [0,60]
x = np.random.uniform(0,60,N)
y = np.random.uniform(0,60,N)

# Añadir la condición de encuentro
# Color rosa los puntos donde se encuentran y de morado en donde NO se encuentran
encuentro = np.where(abs(x-y)<=10, 'pink','purple')

# Calcular probabilidad de que se encuentren
k = 0 # contador
for e in encuentro:# e = elemento
  if e == 'pink': # favorables
       k += 1 # añade al contador
print('La probabilidad del evento A es {}'.format(k/N))

# Gráfica
plt.figure(figsize=(5,5))
plt.scatter(x,y,c=encuentro, s=2)
plt.show()
```

La región en donde se encuentran las personas $A$ y $B$ es en la $\color{pink}{rosa}$ y donde no es la $\color{purple}{morada}$
